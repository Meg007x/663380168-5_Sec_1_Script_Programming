# **1. Lab 6.1: Simple Calculator Functions**

```python
# lab6_1.py

def add(a, b):
    """Adds two numbers and returns the result."""
    return a + b

def subtract(a, b):
    """Subtracts the second number from the first and returns the result."""
    return a - b

def multiply(a, b):
    """Multiplies two numbers and returns the result."""
    return a * b

def divide(a, b):
    """Divides the first number by the second. Handles division by zero."""
    if b == 0:
        return "Error: Division by zero"
    return a / b

def power(base, exponent=2):
    """Calculates the power of a base number. Default exponent is 2 (square)."""
    return base ** exponent

def main():
    while True:
        print("\n=== Simple Calculator ===")
        print("1. Add (+)")
        print("2. Subtract (-)")
        print("3. Multiply (*)")
        print("4. Divide (/)")
        print("5. Power (^)")
        print("0. Exit")

        choice = input("Select an operation (0-5): ")

        if choice == '0':
            print("Exiting calculator. Goodbye!")
            break

        if choice in ['1', '2', '3', '4']:
            try:
                num1 = float(input("Enter first number: "))
                num2 = float(input("Enter second number: "))
            except ValueError:
                print("Invalid input! Please enter numeric values.")
                continue

            if choice == '1':
                print(f"Result: {add(num1, num2)}")
            elif choice == '2':
                print(f"Result: {subtract(num1, num2)}")
            elif choice == '3':
                print(f"Result: {multiply(num1, num2)}")
            elif choice == '4':
                print(f"Result: {divide(num1, num2)}")

        elif choice == '5':
            try:
                base = float(input("Enter base number: "))
                exp_input = input("Enter exponent (press Enter for default 2): ")
                if exp_input.strip() == "":
                    print(f"Result (Squared): {power(base)}")
                else:
                    exp = float(exp_input)
                    print(f"Result: {power(base, exp)}")
            except ValueError:
                print("Invalid input! Please enter numeric values.")
        else:
            print("Invalid choice! Please select 0-5.")

if __name__ == "__main__":
    main()
```

# **2. Lab 6.2: Custom Module & Standard Library**

**ไฟล์ที่ 1: my_utils.py**

```python
%%writefile my_utils.py
# my_utils.py

def greet(name):
    """Prints a friendly greeting message."""
    print(f"Hello, {name}!")

def is_prime(number):
    """Returns True if the number is prime, False otherwise."""
    if number <= 1:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True
```

**ไฟล์ที่ 2: main_program.py**

```python
# main_program.py
import my_utils
import math
import random

def main():
    print("--- Testing Custom Module (my_utils) ---")
    # Call greet function
    my_utils.greet("Alice")

    # Test is_prime function
    test_numbers = [7, 10, 13, 1]
    for num in test_numbers:
        print(f"Is {num} prime? {my_utils.is_prime(num)}")

    print("\n--- Testing Standard Libraries ---")
    # Using math module
    num_to_sqrt = 64
    print(f"Square root of {num_to_sqrt} using math.sqrt: {math.sqrt(num_to_sqrt)}")

    # Using random module
    rand_val = random.randint(1, 100)
    print(f"Random number between 1 and 100: {rand_val}")

if __name__ == "__main__":
    main()
```

# **3. Activity 6.1: Refactoring Code into Functions**

```python
# activity6_1.py

def is_positive_negative_zero(num):
    """Classifies a number as Positive, Negative, or Zero."""
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

def is_even_odd(num):
    """Determines whether an integer is Even or Odd."""
    if num % 1 == 0:  # Check if it's a whole number
        if int(num) % 2 == 0:
            return "Even"
        else:
            return "Odd"
    return "N/A (Not an Integer)"

def main():
    print("=== Number Classifier (Refactored) ===")
    try:
        user_input = float(input("Enter a number to classify: "))

        sign_result = is_positive_negative_zero(user_input)
        parity_result = is_even_odd(user_input)

        print(f"\nClassification Results for {user_input}:")
        print(f"- Sign: {sign_result}")
        print(f"- Parity: {parity_result}")

    except ValueError:
        print("Invalid input! Please enter a valid number.")

if __name__ == "__main__":
    main()
```
